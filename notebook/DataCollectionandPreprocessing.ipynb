```python
# introduce some missing values
def introduce_missing_values(df, missing_rate=0.1):
    synthetic_data__copy = df.copy()
    total_cells = synthetic_data__copy.size
    num_missing = int(total_cells * missing_rate)

    for _ in range(num_missing):
        i = np.random.randint(0, synthetic_data__copy.shape[0])
        j = np.random.randint(0, synthetic_data__copy.shape[1])
        synthetic_data__copy.iat[i, j] = np.nan

    return synthetic_data__copy


synthetic_df_copy = introduce_missing_values(synthetic_data)

synthetic_df_copy
```

## Step 7 Cleaning Rules

```python
# define a function that does the cleaning
column_rename = {
    "Unit Sold": "Quantity",
    
}
def clean_df(df, rename_map=column_rename):
    """
    Clean the merged_df DataFrame by:
    1. Removing duplicate rows
    2. Renaming columns
    3. Dropping rows with missing values
    """
    # Step 1: Remove duplicate rows
    df_cleaned = df.drop_duplicates()

    # Step 2: Rename columns if a mapping is provided
    if rename_map:
        df_cleaned = df_cleaned.rename(columns=rename_map)

    # Step 3: Drop rows with any missing values
    merged_df_cleaned = df_cleaned.dropna()

    return merged_df_cleaned
    
cleaned_df = clean_df(merged_sales_df)
```

## Step 8 : Transformations

```python
cleaned_df.info()
```

```python
# convert date columns to the right type
date_columns = ["Order Date", "Ship Date"]

for col in date_columns:
    cleaned_df[col] = pd.to_datetime(cleaned_df[col], errors='coerce')
    

cleaned_df.info()

cleaned_df.head()
```

```python
# transform my postal code to numeric

cleaned_df['Postal Code'] = pd.to_numeric(cleaned_df['Postal Code'], errors='coerce')

cleaned_df.info()
```

## Step 9 Feature Engineering

```python
# add wait time in days from when the user orders to when it is shipped in days

cleaned_df["Wait Time"] = (cleaned_df["Ship Date"] - cleaned_df["Order Date"]).dt.days

cleaned_df["Wait Time"]
```

## Step 10 Mini- Aggregation

```python
# group the item type by wait time 
# - shows how many items of each type fall under each wait time

group1 = cleaned_df.groupby('Wait Time')['Item Type'].value_counts()
print(group1)
```

```python
# group mean revenue by sales channel

group2 = cleaned_df.groupby('Sales Channel')['Total Revenue'].mean()
group2
```

```python
# group iten type by country and revenue

group3 = cleaned_df.groupby(['Item Type', 'Country'])['Total Revenue'].mean().unstack()
group3
```

```python
# save to a dictionary the revenue of each country
revenue_by_country = cleaned_df.groupby('Country')['Total Revenue'].sum().to_dict()

revenue_by_country
```

### 🛠️ Notes on Creation
- **Synthetic** fields like `Coupon Code`, `City`, and `Customer ID` were generated using tools like Faker to simulate realistic data.
    - The `Customer ID` was created by concatenating a fixed prefix with a randomly generated integer between 100000 and 200000
    - `City` and `Postal Code` use Faker to generate realistic but fictitious location data
    - `Coupon Description` is a formatted string combining a discount type and product name.
    - The `coupon code` is generated by randomly selecting uppercase letters and digits to form a string of a specified length, creating a unique alphanumeric promotional code

- **Derived** field `Wait Time` was derived from date subtraction.
- All types were standardized during cleaning, with dates converted to `datetime` and numeric fields cast to `float` or `integer`.
